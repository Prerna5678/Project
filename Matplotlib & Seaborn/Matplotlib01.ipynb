```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Housing.csv')
```

```python
df
```

```python
# Line Chart

x = df['price']
y = df['area']
plt.plot(x, y)
plt.xlabel('Price')
plt.ylabel('Area')
plt.title('Price vs Area')
plt.show()
```

```python
# Bar Chart

x = df['prefarea']
y = df['furnishingstatus']
plt.bar(x, y)
plt.xlabel('prefarea')
plt.ylabel('furnishingstatus')
plt.title('prefarea vs furnishingstatus')
plt.show()
```

```python
#  Histogram

x = df['price']
y = df['area']
plt.hist(x, bins=20)
plt.xlabel('Price')
plt.ylabel('Area')
plt.title('Price vs Area')
plt.show()
```

```python
# Scatter Plot

x = df['price']
y = df['area']
plt.scatter(x, y)
plt.xlabel('Price')
plt.ylabel('Area')
plt.title('Price vs Area')
plt.show()
```

```python
# Pie Chart

labels = ['Furnished', 'Semi-Furnished', 'Unfurnished']
data = [52, 35, 13]  # Example values

plt.pie(data, labels=labels, autopct='%1.1f%%')
plt.title("Housing Furnishing Status")
plt.show()
```

```python
#  Box Plot

x = df['furnishingstatus']
plt.boxplot(x=df['price'], labels=['Price'])
plt.xlabel('Furnishing Status')
plt.ylabel('Price')
plt.title('Furnishing Status')
plt.show()
```

```python
# Heatmap


# Select numerical columns
housing_data = df[['price', 'area', 'bedrooms', 'bathrooms', 'parking']]

# Calculate correlation matrix
corr_matrix = housing_data.corr()

# Create heatmap
plt.imshow(corr_matrix, cmap='viridis', interpolation='nearest')

plt.colorbar()

plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=45)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)

plt.title('Housing Data Correlation Heatmap')
plt.tight_layout()

plt.show()
```
