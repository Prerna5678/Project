```python
# Slicing in Numpy Array
```

```python
import numpy as np
```

```python
# 1d array
array1 = np.array([10, 20, 30, 40, 50, 60, 70])
print(array1)
```

```python
# start indexing and stop indexing
print(array1[1:3])
```

```python
# start indexing and stop indexing and step (gap) is two here
print(array1[1:6:2])
```

```python
print(array1[-1:-3:-1])
```

```python
# :: start and end value not mentioned and 2 means gap/step
print(array1[::2])
```

```python
# reverse
print(array1[::-1])
```

```python
# 2d array (4x3 matrix = 12 toatal)
#                   0   1   2 --> column
array2 = np.array([[10, 20, 30], # 0 rows
                  [40, 50, 60], # 1 rows
                  [70, 80, 90], # 2 rows
                  [100, 110, 120]]) # 3 rows
print(array2)
```

```python
# rows first after comma column come
print(array2[1,])
```

```python
print(array2[:,1])
```

```python
# row and column range
print(array2[1:3 , 1:3])
```

```python
print(array2[0:2 , 0:2 ])
```

```python
print(array2[1:3 , ])
```

```python
print(array2[:, 1:3])
```

```python
print(array2[1:3, 1])
```

```python
# row and column after :1 means end value
print(array2[1:3, :1])
```

```python
# row and column after 1: means start value
print(array2[1:3, 1:])
```
