```
def preprocess_image(image_path):
    img = load_img(image_path, target_size=(299, 299))
    img = img_to_array(img)
    img = np.expand_dims(img, axis=0)
    img = tf.keras.applications.inception_v3.preprocess_input(img)
    return img

def extract_image_features(model, image_path):
    img = preprocess_image(image_path)
    features = model.predict(img, verbose=0)
    return features

inception_v3_model = InceptionV3(weights = 'imagenet', input_shape=(299, 299, 3))
inception_v3_model.layers.pop()
inception_v3_model = Model(inputs=inception_v3_model.inputs, outputs=inception_v3_model.layers[-2].output)
```

```
train_image_features, val_image_features, test_image_features = {}, {}, {}  # A Dictionary to store image features with their corresponding IDs

# Filter out directories from the list of items in images_directory
image_files = [f for f in os.listdir(images_directory) if os.path.isfile(os.path.join(images_directory, f))]

pbar = tqdm_notebook(total=len(image_files), position=0, leave=True, colour='green')

for image_id in image_files:
    image_path = os.path.join(images_directory, image_id)
    try:
        image_features = extract_image_features(inception_v3_model, image_path) # Extracting features

        if image_id in train_image_ids:
            train_image_features[image_id] = image_features.flatten()  # Flattening the features
            pbar.update(1)

        elif image_id in val_image_ids:
            val_image_features[image_id] = image_features.flatten()  # Flattening the features
            pbar.update(1)

        elif image_id in test_image_ids:
            test_image_features[image_id] = image_features.flatten()  # Flattening the features
            pbar.update(1)

    except Exception as e:
        print(f"Could not process image {image_id}: {e}")
        pbar.update(1)


pbar.close()

print(f"Extracted {len(train_image_features)} training image features.")
print(f"Extracted {len(val_image_features)} validation image features.")
print(f"Extracted {len(test_image_features)} test image features.")
```

```
# Determine the maximum caption length
max_caption_length = max(len(caption.split()) for caption in cleaned_captions)
print(f"Maximum caption length: {max_caption_length}")

# Create word-to-index and index-to-word mappings
word_to_index = tokenizer.word_index
index_to_word = {index: word for word, index in word_to_index.items()}

# Convert captions to sequences of integers
caption_sequences = tokenizer.texts_to_sequences(cleaned_captions)

# Pad sequences to the maximum length
padded_sequences = pad_sequences(caption_sequences, maxlen=max_caption_length, padding='post')

print(f"Shape of padded sequences: {padded_sequences.shape}")
```

```
def data_generator(captions, image_features, tokenizer, max_caption_length, batch_size):
    num_samples = len(captions)
    # image_ids = list(image_features.keys()) # No longer needed for shuffling entire dataset

    # Shuffle captions to shuffle image_ids and captions together
    np.random.shuffle(captions)

    for start_idx in range(0, num_samples, batch_size):
        end_idx = min(start_idx + batch_size, num_samples)
        batch_captions = captions[start_idx:end_idx]

        X_images, X_captions, y = [], [], []
        for caption in batch_captions:
            image_id, caption_text = caption.split('\t')
            caption_text = caption_text.rstrip('\n')
            seq = tokenizer.texts_to_sequences([caption_text])[0] # Tokenizing the caption
            for i in range(1, len(seq)):
                in_seq, out_seq = seq[:i], seq[i] # X_caption, Y
                in_seq = pad_sequences([in_seq], maxlen=max_caption_length)[0]
                out_seq = to_categorical([out_seq], num_classes=vocab_size)[0]
                # Append individual samples to lists
                X_images.append(image_features[image_id])
                X_captions.append(in_seq)
                y.append(out_seq)

        # Add print statements to inspect the yielded data
        print("Yielding batch:")
        print(f"  Type of yielded object: {type(((np.array(X_images), np.array(X_captions)), np.array(y)))})") # Added print for the full yielded object type
        print(f"  Type of inputs: {type((np.array(X_images), np.array(X_captions)))})") # Changed to check tuple type
        print(f"  Type of targets: {type(np.array(y))}")
        print(f"  Shape of X_images: {np.array(X_images).shape}")
        print(f"  Shape of X_captions: {np.array(X_captions).shape}")
        print(f"  Shape of y: {np.array(y).shape}")

        # Yield batches
        yield (np.array(X_images), np.array(X_captions)), np.array(y) # Yielding a tuple for inputs

# Determine the maximum caption length and add 1 for the 'end' token
max_caption_length = max(len(caption.split()) for caption in cleaned_captions) + 1

cnn_output_dim = inception_v3_model.output_shape[1] # 2048

batch_size_train = 270
batch_size_val = 150

# The data_generator now yields batches, so output_shapes should reflect the batch dimension.
# However, tf.data.Dataset.from_generator infers the batch dimension automatically.
# We just need to define the shapes of a single element within a batch.
# The shapes defined in AOEzJmDlEU9q are correct for the element shape.


train_data_generator = data_generator(train_captions, train_image_features, tokenizer, max_caption_length -1, batch_size_train)
val_data_generator = data_generator(val_captions, val_image_features, tokenizer, max_caption_length -1, batch_size_val)
```

```
def build_model(vocab_size, max_caption_length, cnn_output_dim):
    input_image = Input(shape=(cnn_output_dim,), name='Features_Input')
    fe1 = BatchNormalization()(input_image)
    fe2 = Dense(256, activation='relu')(fe1) # Adding a Dense layer to the CNN output to match the decoder output size
    fe3 = BatchNormalization()(fe2)

    input_caption = Input(shape=(max_caption_length -1,), name='Sequence_Input') # Corrected input shape
    se1 = Embedding(vocab_size, 256, mask_zero=True)(input_caption)
    se2 = LSTM(256)(se1)

    decoder1 = add([fe3, se2])
    decoder2 = Dense(256, activation='relu')(decoder1)
    outputs = Dense(vocab_size, activation='softmax', name='Output_Layer')(decoder2)

    model = Model(inputs=[input_image, input_caption], outputs=outputs, name='Image_Captioning')
    return model

caption_model = build_model(vocab_size, max_caption_length, cnn_output_dim)

optimizer = Adam(learning_rate=0.01, clipnorm=1.0)
caption_model.compile(loss='categorical_crossentropy', optimizer=optimizer)

caption_model.summary()
```

Based on the code in cell `4kIn1aONExkW`, the expected structure for the `tf.data.Dataset.from_generator` is:

*   **`output_types`**: `((tf.float32, tf.int32), tf.float32)`
    *   This indicates that each element yielded by the generator should be a tuple.
    *   The first element of the tuple is another tuple `(tf.float32, tf.int32)`, representing the inputs to the model.
        *   `tf.float32` is for the image features.
        *   `tf.int32` is for the caption sequences (since they are integer IDs).
    *   The second element of the main tuple is `tf.float32`, representing the target (the one-hot encoded next word).

*   **`output_shapes`**: `(((None, cnn_output_dim), (None, max_caption_length -1)), (None, vocab_size,))`
    *   This indicates the shapes of the tensors within the structure defined by `output_types`.
    *   `((None, cnn_output_dim), (None, max_caption_length -1))` corresponds to the shapes of the inputs:
        *   `(None, cnn_output_dim)` for the image features (`None` for batch size, `cnn_output_dim` is 2048).
        *   `(None, max_caption_length -1))` for the caption sequences (`None` for batch size, `max_caption_length - 1` is 37).
    *   `(None, vocab_size,)` corresponds to the shape of the target (`None` for batch size, `vocab_size` is 680).

So, the generator should yield elements that match this structure: a tuple where the first element is a tuple of two tensors (image features batch and caption sequences batch), and the second element is a tensor (target words batch).
