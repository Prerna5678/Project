```python
'''
Copy vs View in NumPy

In NumPy, arrays can be copied in two ways:

View
Copy

Understanding this is very important because sometimes changing one array also changes another array.
'''
```

```python
import numpy as np
```

```python
'''
1. View in NumPy

A view shares the same data with the original array.

If you change the view, the original array also changes.

syntax:- view()
'''

a = np.array([1,2,3,4])

b = a.view()

b[0] = 100

print("Original Array:", a)
print("View Array:", b)
```

```python
'''
2. Copy in NumPy

A copy creates completely separate array.

Changes in copied array do NOT affect original array.

Use:
copy()
'''

import numpy as np

a = np.array([1,2,3,4])

b = a.copy()

b[0] = 100

print("Original Array:", a)
print("Copied Array:", b)
```

```python
'''
Check Base Attribute

The base attribute tells whether array owns data or not.
'''
# Means b depends on original array.

a = np.array([1,2,3])

b = a.view()

print(b.base)
```

```python
'''
Difference Between Copy and View

| Feature                 | View | Copy            |
| ----------------------- | ---- | --------------- |
| Shares memory           | Yes  | No              |
| Changes affect original | Yes  | No              |
| Faster                  | Yes  | Slightly slower |
| Extra memory needed     | No   | Yes             |


Important Points
view() → linked array
copy() → independent array
View saves memory
Copy is safer when editing data
'''
```

```python
# Practical Example

import numpy as np

a = np.array([10,20,30])

# view
v = a.view()

# copy
c = a.copy()

a[0] = 999

print("Original:", a)
print("View:", v)
print("Copy:", c)
```
