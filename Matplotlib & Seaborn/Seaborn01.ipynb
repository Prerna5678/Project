```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Housing.csv')
```

```python
df
```

```python
# Line plot
# Sort data by area for a proper line plot
df_sorted = df.sort_values('price')

sns.lineplot(
    x='price',
    y='area',
    data=df_sorted
)

plt.xlabel('Price')
plt.ylabel('Area')
plt.title('Area vs Price')
plt.show()
```

```python
avg_price = df.groupby('bedrooms')['price'].mean().reset_index()

sns.lineplot(
    x='bedrooms',
    y='price',
    data=avg_price
)

plt.xlabel('Bedrooms')
plt.ylabel('Average Price')
plt.title('Average Price by Bedrooms')
plt.show()
```

```python
# Scatter Plot
sns.scatterplot(
    x='area',
    y='price',
    data=df
)

plt.xlabel('Area')
plt.ylabel('Price')
plt.title('Area vs Price')
plt.show()
```

```python
# Box Plot

sns.boxplot(y='price', data=df)

plt.ylabel('Price')
plt.title('House Price Distribution')
plt.show()
```

```python
# Violin Plot

sns.violinplot(y='price', data=df)

plt.ylabel('Price')
plt.title('House Price Distribution')
plt.show()
```

```python
# Swarm plot

sns.swarmplot(
    x='bedrooms',
    y='price',
    data=df
)

plt.xlabel('Bedrooms')
plt.ylabel('Price')
plt.title('Bedrooms vs House Price')
plt.show()
```

```python
# Bar plot

avg_price = df.groupby('bedrooms')['price'].mean().reset_index()

sns.barplot(x='bedrooms', y='price', data=avg_price)

plt.xlabel('Bedrooms')
plt.ylabel('Average Price')
plt.title('Average House Price by Bedrooms')
plt.show()
```

```python
# Point plot

sns.pointplot(
    x='bathrooms',
    y='price',
    data=df
)

plt.xlabel('Bathrooms')
plt.ylabel('Average Price')
plt.title('Average House Price by Bathrooms')
plt.show()
```
