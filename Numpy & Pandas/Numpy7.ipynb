```python
'''
Statistical operator on 1d array
1. max()
2. min()
3. sum()
4. mean()
5. median()
6. prod()
7. var()
8. std()
'''
```

```python
import numpy as np
```

```python
# 1. max()

x = np.array([4, 2, 3, 9, 7])
y = np.max(x)
print(y)
```

```python
# 2. min()

x = np.array([4, 2, 3, 9, 7])
y = np.min(x)
print(y)
```

```python
# 3. sum()

x = np.array([4, 2, 3, 9, 7])
y = np.sum(x)
print(y)
```

```python
# 4. mean()

x = np.array([4, 2, 3, 9, 7])
y = np.mean(x)
print(y)
```

```python
# 5. median()

x = np.array([4, 2, 3, 9, 7])
y = np.median(x)
print(y)
```

```python
# 6.prod()

x = np.array([4, 2, 3, 9, 7])
y = np.prod(x)
print(y)
```

```python
# 7. var()

x = np.array([4, 2, 3, 9, 7])
y = np.var(x)
print(y)
```

```python
# 8. std()

x = np.array([4, 2, 3, 9, 7])
y = np.std(x)
print(y)
```
