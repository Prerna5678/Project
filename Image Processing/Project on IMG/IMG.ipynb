````
# Basic libraries
import os
import pickle
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import warnings
warnings.filterwarnings('ignore')
from math import ceil
from collections import defaultdict
from tqdm.notebook import tqdm        # Progress bar library for Jupyter Notebook

# Deep learning framework for building and training models
import tensorflow as tf
## Pre-trained model for image feature extraction
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input
from tensorflow.keras.preprocessing.image import load_img, img_to_array

## Tokenizer class for captions tokenization
from tensorflow.keras.preprocessing.text import Tokenizer

## Function for padding sequences to a specific length
from tensorflow.keras.preprocessing.sequence import pad_sequences

## Class for defining Keras models
from tensorflow.keras.models import Model
from tensorflow.keras.utils import to_categorical, plot_model
from tensorflow.keras.layers import Input, Dense, LSTM, Embedding, Dropout, concatenate, Bidirectional, Dot, Activation, RepeatVector, Multiply, Lambda

# For checking score
from nltk.translate.bleu_score import corpus_bleu
````

````
# We are going to use pretraind vgg model
# Load the vgg16 model
model = VGG16()

# Restructuring the model to remove the last classification layer, this will give us access to the output features of the model
model = Model(inputs=model.inputs, outputs=model.layers[-2].output)

# Printing the model summary
print(model.summary())
````

````
# Initialize an empty dictionary to store image features
image_features = {}

# Define the directory path where images are located
# Corrected: Point to the subdirectory containing actual images
img_dir = os.path.join(INPUT_DIR, 'Image') # Assuming images are in 'Image' subdirectory

# Loop through each image in the directory
# Filter for .jpg files to ensure we only process images
for img_name in tqdm([f for f in os.listdir(img_dir) if f.lower().endswith('.jpg')]):
    # Load the image from file
    img_path = os.path.join(img_dir, img_name)
    image = load_img(img_path, target_size=(224, 224))
    # Convert image pixels to a numpy array
    image = img_to_array(image)
    # Reshape the data for the model
    image = image.reshape((1, image.shape[0], image.shape[1], image.shape[2]))
    # Preprocess the image for ResNet50
    image = preprocess_input(image)
    # Extract features using the pre-trained ResNet50 model
    image_feature = model.predict(image, verbose=0)
    # Get the image ID by removing the file extension
    image_id = img_name.split('.')[0]
    # Store the extracted feature in the dictionary with the image ID as the key
    image_features[image_id] = image_feature
````

````
import os
from collections import defaultdict
from tqdm.notebook import tqdm

# Define the directory path where images are located (already defined as INPUT_DIR)
# Define the path to the Caption.txt file
captions_file_path = os.path.join(INPUT_DIR, 'Caption.txt')

# Check if the captions file exists
if not os.path.exists(captions_file_path):
    raise FileNotFoundError(f"❌ Caption.txt not found at {captions_file_path}. Please ensure it is uploaded.")

print(f"✔ Found captions file: {captions_file_path}")

# Load captions_doc from 'Caption.txt'
with open(captions_file_path, 'r') as file:
    next(file) # Skip the header line if present
    captions_doc = file.read()

# Re-create image_to_captions_mapping
image_to_captions_mapping = defaultdict(list)
for line in tqdm(captions_doc.split('\n')):
    if line:
        parts = line.split(',')
        if len(parts) >= 2:
            image_id = parts[0].split('.')[0] # Remove .jpg extension
            caption = ','.join(parts[1:]) # Rejoin parts in case caption had commas
            image_to_captions_mapping[image_id].append(caption.strip())

# Re-clean the captions function (already defined in the notebook, but including here for completeness)
def clean(mapping):
    for key, captions in mapping.items():
        for i in range(len(captions)):
            caption = captions[i]
            caption = caption.lower()
            caption = ''.join(char for char in caption if char.isalpha() or char.isspace())
            caption = caption.replace('\s+', ' ')
            caption = 'startseq ' + ' '.join([word for word in caption.split() if len(word) > 1]) + ' endseq'
            captions[i] = caption
clean(image_to_captions_mapping)

# Filter loaded_image_features to keep only images that have captions
# Assuming loaded_image_features is available from a previous cell
# If not, it needs to be loaded first, which is done in 6gORWJqoRCjf

# Reload image_features from pickle file (if not already loaded or for consistency)
pickle_file_path = os.path.join(OUTPUT_DIR, 'img_features.pkl')
with open(pickle_file_path, 'rb') as file:
    loaded_image_features = pickle.load(file)

all_image_ids = list(image_to_captions_mapping.keys())
filtered_image_features = {img_id: feature for img_id, feature in loaded_image_features.items() if img_id in all_image_ids}

# Filter image_to_captions_mapping to keep only captions for images that have features
filtered_image_to_captions_mapping = {img_id: captions for img_id, captions in image_to_captions_mapping.items() if img_id in filtered_image_features}

# Update the main variables
image_features = filtered_image_features
image_to_captions_mapping = filtered_image_to_captions_mapping

# Create a list of all captions
all_captions = [caption for captions in image_to_captions_mapping.values() for caption in captions]

print("✔ Total captions after cleaning and filtering:", len(all_captions))

# Initialize and fit tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(all_captions)

# Save the tokenizer
with open(os.path.join(OUTPUT_DIR, 'tokenizer.pkl'), 'wb') as tokenizer_file:
    pickle.dump(tokenizer, tokenizer_file)

# Load the tokenizer (for immediate use after saving)
with open(os.path.join(OUTPUT_DIR, 'tokenizer.pkl'), 'rb') as tokenizer_file:
    tokenizer = pickle.load(tokenizer_file)

# Calculate max_caption_length and vocab_size
max_caption_length = max(len(tokenizer.texts_to_sequences([caption])[0]) for caption in all_captions)
vocab_size = len(tokenizer.word_index) + 1

print("Vocabulary Size:", vocab_size)
print("Maximum Caption Length:", max_caption_length)

# Prepare training data (X1, X2, y)
X1, X2, y = [], [], []

for img_id, captions in tqdm(image_to_captions_mapping.items()):
    for caption in captions:
        # Convert caption to sequence of integers
        seq = tokenizer.texts_to_sequences([caption])[0]
        for i in range(1, len(seq)):
            # Split sequence into input and output parts
            in_seq, out_seq = seq[:i], seq[i]
            # Pad input sequence to max_caption_length
            in_seq = pad_sequences([in_seq], maxlen=max_caption_length)[0]
            # Convert output sequence to one-hot encoding
            out_seq = to_categorical([out_seq], num_classes=vocab_size)[0]

            # Store the data
            X1.append(image_features[img_id])
            X2.append(in_seq)
            y.append(out_seq)

# Convert lists to numpy arrays
X1 = np.array(X1)
X2 = np.array(X2)
y = np.array(y)

print(f"Shape of X1 (image features): {X1.shape}")
print(f"Shape of X2 (caption input sequences): {X2.shape}")
print(f"Shape of y (caption output sequences): {y.shape}")
````

````
# Creating a List of All Captions
all_captions = [caption for captions in image_to_captions_mapping.values() for caption in captions]
````

````
# Reshape X1 for the model input
X1_reshaped = X1.reshape(X1.shape[0], -1) # Flatten the second dimension (1) if present
````

````
epochs = 20

# Convert NumPy arrays to TensorFlow tensors with correct dtypes
X1_tf = tf.convert_to_tensor(X1_reshaped, dtype=tf.float32)
X2_tf = tf.convert_to_tensor(X2, dtype=tf.int32)
y_tf = tf.convert_to_tensor(y, dtype=tf.float32)

# Re-create the model with explicit dtypes in Input layers
image_features_input = Input(shape=(4096,), dtype=tf.float32, name='image_features_input')
img_feature_encoder = Dense(256, activation='relu')(image_features_input)

caption_input = Input(shape=(max_caption_length,), dtype=tf.int32, name='caption_input')
caption_embedding = Embedding(vocab_size, 256, mask_zero=True)(caption_input)
caption_lstm = LSTM(256)(caption_embedding)

decoder_output = concatenate([img_feature_encoder, caption_lstm])
decoder_output = Dense(256, activation='relu')(decoder_output)
model_output = Dense(vocab_size, activation='softmax')(decoder_output)

model = Model(inputs=[image_features_input, caption_input], outputs=model_output)
model.compile(optimizer='adam', loss='categorical_crossentropy')

# Create a tf.data.Dataset from tensors
dataset = tf.data.Dataset.from_tensor_slices(( (X1_tf, X2_tf), y_tf ))

# Shuffle and batch the dataset
buffer_size = len(X1_tf) # Use full dataset size for shuffling
dataset = dataset.shuffle(buffer_size=buffer_size).batch(batch_size)

# Train the model
model.fit(
    dataset,
    epochs=epochs,
    verbose=1
)

print(f"Model training completed for {epochs} epochs.")
````

````
# Data generator function
def data_generator(data_keys, image_to_captions_mapping, features, tokenizer, max_caption_length, vocab_size, batch_size):
    # Lists to store batch data
    X1_batch, X2_batch, y_batch = [], [], []
    # Counter for the current batch size
    batch_count = 0

    while True:
        # Loop through each image in the current batch
        for image_id in data_keys:
            # Get the captions associated with the current image
            captions = image_to_captions_mapping[image_id]

            # Loop through each caption for the current image
            for caption in captions:
                # Convert the caption to a sequence of token IDs
                caption_seq = tokenizer.texts_to_sequences([caption])[0]

                # Loop through the tokens in the caption sequence
                for i in range(1, len(caption_seq)):
                    # Split the sequence into input and output pairs
                    in_seq, out_seq = caption_seq[:i], caption_seq[i]

                    # Pad the input sequence to the specified maximum caption length
                    in_seq = pad_sequences([in_seq], maxlen=max_caption_length)[0]

                    # Convert the output sequence to one-hot encoded format
                    out_seq = to_categorical([out_seq], num_classes=vocab_size)[0]

                    # Append data to batch lists
                    X1_batch.append(features[image_id][0])  # Image features
                    X2_batch.append(in_seq)  # Input sequence
                    y_batch.append(out_seq)  # Output sequence

                    # Increase the batch counter
                    batch_count += 1

                    # If the batch is complete, yield the batch and reset lists and counter
                    if batch_count == batch_size:
                        X1_batch, X2_batch, y_batch = np.array(X1_batch), np.array(X2_batch), np.array(y_batch)
                        yield [X1_batch, X2_batch], y_batch
                        X1_batch, X2_batch, y_batch = [], [], []
                        batch_count = 0
````

````
# Encoder model
inputs1 = Input(shape=(4096,))
fe1 = Dropout(0.5)(inputs1)
fe2 = Dense(256, activation='relu')(fe1)
fe2_projected = RepeatVector(max_caption_length)(fe2)
fe2_projected = Bidirectional(LSTM(256, return_sequences=True))(fe2_projected)

# Sequence feature layers
inputs2 = Input(shape=(max_caption_length,))
se1 = Embedding(vocab_size, 256, mask_zero=True)(inputs2)
se2 = Dropout(0.5)(se1)
se3 = Bidirectional(LSTM(256, return_sequences=True))(se2)

# Apply attention mechanism using Dot product
attention = Dot(axes=[2, 2])([fe2_projected, se3])  # Calculate attention scores

# Softmax attention scores
attention_scores = Activation('softmax')(attention)

# Apply attention scores to sequence embeddings
attention_context = Lambda(lambda x: tf.einsum('ijk,ijl->ikl', x[0], x[1]))([attention_scores, se3])

# Sum the attended sequence embeddings along the time axis
context_vector = Lambda(lambda x: tf.reduce_sum(x, axis=1))(attention_context) # Wrapped tf.reduce_sum in Lambda layer

# Decoder model
decoder_input = concatenate([context_vector, fe2], axis=-1)
decoder1 = Dense(256, activation='relu')(decoder_input)
outputs = Dense(vocab_size, activation='softmax')(decoder1)

# Create the model
model = Model(inputs=[inputs1, inputs2], outputs=outputs)
model.compile(loss='categorical_crossentropy', optimizer='adam')

# Visualize the model
plot_model(model, show_shapes=True)
````

## Define Data Generator

6. Yield the batch as a tuple: `([batch_X1, batch_X2], batch_y)`.

**Reasoning**:
I need to generate the Python code for the `data_generator` function as described in the previous instructions. This function will create batches of image features, caption input sequences, and caption output sequences for model training.



````
import math

# Define the data generator function
def data_generator(X1, X2, y, batch_size):
    num_samples = len(X1)
    num_batches = math.ceil(num_samples / batch_size)

    # Infinite loop to yield batches continuously
    while True:
        for i in range(num_batches):
            start_index = i * batch_size
            end_index = min((i + 1) * batch_size, num_samples)

            batch_X1 = X1[start_index:end_index]
            batch_X2 = X2[start_index:end_index]
            batch_y = y[start_index:end_index]

            # Reshape batch_X1 for the model's input if it's currently (batch_size, 1, 4096)
            # The kernel state for X1 shows (2783, 1, 4096), so we keep the (batch_size, 1, 4096) shape.

            yield ([batch_X1, batch_X2], batch_y)

print("Data generator function defined.")
````

````
def get_word_from_index(index, tokenizer):
    return next((word for word, idx in tokenizer.word_index.items() if idx == index), None)
````

````


def predict_caption(model, photo_feature, tokenizer, max_length):
    """
    Generate caption for an image using greedy search.
    photo_feature: numpy array shape (4096,) or (1,4096)
    """

    # Ensure correct shape
    if photo_feature.ndim == 1:
        photo_feature = photo_feature.reshape(1, 4096)

    # Start the sequence
    in_text = "startseq"

    for _ in range(max_length):

        # Convert words → integer sequence
        sequence = tokenizer.texts_to_sequences([in_text])[0]

        # Pad sequence
        sequence = tf.keras.preprocessing.sequence.pad_sequences(
            [sequence],
            maxlen=max_length
        )

        # Predict next word
        yhat = model.predict([photo_feature, sequence], verbose=0)
        yhat = yhat.argmax()   # get index of best word

        # Convert index → word
        word = tokenizer.index_word.get(yhat)

        # If word not found
        if word is None:
            break

        # Append word
        in_text += ' ' + word

        # If end token predicted
        if word == "endseq":
            break

    # Remove start/end tokens
    final_caption = (
        in_text.replace("startseq", "")
               .replace("endseq", "")
               .strip()
    )

    return final_caption
````
