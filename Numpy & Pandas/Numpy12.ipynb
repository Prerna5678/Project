```python
'''
Linear Algebra Basics in NumPy

NumPy provides a special module called:

np.linalg

It is used for matrix operations like:-
Matrix addition
Matrix multiplication
Determinant
Inverse
Transpose
Eigen values
Dot product
'''
```

```python
import numpy as np
```

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A)
print(B)
```

```python
# 2. Matrix Addition

result = A + B

print(result)
```

```python
# 3. Matrix Subtraction

result = A - B
print(result)
```

```python
'''
4. Matrix Multiplication

Using dot():
'''

result = np.dot(A, B)
print(result)
```

```python
# Or using @

result = A @ B

print(result)
```

```python
# 5. Transpose of Matrix

result = A.T

print(result)
```

```python
# 6. Determinant of Matrix
# Determinant tells important properties of matrix

det_A = np.linalg.det(A)    
print(det_A)
```

```python
# 7. Inverse of Matrix
# Inverse exists only if determinant ≠ 0.

inv_A = np.linalg.inv(A)
print(inv_A)
```

```python
# 8. Eigen Values and Eigen Vectors

values, vectors = np.linalg.eig(A)

print("Eigen Values:")
print(values)

print("Eigen Vectors:")
print(vectors)
```

```python
# 9. Dot Product
'''
Calculation:

1×4+2×5+3×6=32
'''
a = np.array([1,2,3])

b = np.array([4,5,6])

result = np.dot(a,b)

print(result)
```

```python
# 10. Identity Matrix

I = np.eye(3)

print(I)
```

```python
'''
Important Functions

| Function          | Meaning               |
| ----------------- | --------------------- |
| `np.dot()`        | Matrix multiplication |
| `A.T`             | Transpose             |
| `np.linalg.det()` | Determinant           |
| `np.linalg.inv()` | Inverse               |
| `np.linalg.eig()` | Eigen values/vectors  |
| `np.eye()`        | Identity matrix       |

'''
```

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print("Addition")
print(A + B)

print("Multiplication")
print(np.dot(A,B))

print("Transpose")
print(A.T)

print("Determinant")
print(np.linalg.det(A))
```
