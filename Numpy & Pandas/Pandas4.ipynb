```python
import pandas as pd
```

```python
# Step 1: Read CSV File

df = pd.read_csv("Housing.csv")
```

```python
df
```

```python
# Step 2: Check Null Values

df.isnull()
# print(df.isnull())
```

```python
# Step 3: Count Null Values Column-wise

df.isnull().sum()
```

```python
# Step 4: Remove Rows with Null Values
'''
Breaking the code
dropna()
drop → remove
na → null/missing values
'''
df.dropna(inplace=True)
```

```python
# Step 5: Fill Null Values with 0
'''
Replace all missing values with 0.
inplace=True
means changes happen in original DataFrame.
'''

df.fillna(0, inplace=True)
```

```python
# Step 6: Fill with Mean Value
'''
Meaning
Missing values in price
are replaced by average price.
'''

df['price'] = df['price'].fillna(df['price'].mean())
```

```python
# Step 7: Check Again

df.isnull().sum()
```

```python
# Most Common Real Use

import pandas as pd

df = pd.read_csv("housing.csv")

# check null values
print(df.isnull().sum())

# fill missing values
df.fillna(0, inplace=True)

# verify
print(df.isnull().sum())
```
