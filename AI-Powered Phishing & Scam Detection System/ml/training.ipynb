```python
df = df.drop_duplicates()
```

```python
# STEP 2: Clean column names
df.columns = df.columns.str.strip()
```

```python
# STEP 3: Handle invalid numerical values
df = df.replace([np.inf, -np.inf], np.nan)
```

```python
print(df.isnull().sum())
```

```python
# Remove rows with missing target labels
df = df.dropna(subset=["Result"])
```

```python
# STEP 4: Remove the row index column
df = df.drop(columns=["index"], errors="ignore")
```

```python
# Verify the dataset
print("\nMissing values:")
print(df.isnull().sum().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("\nTarget distribution:")
print(df["Result"].value_counts())
```

```python
# STEP 5: Separate features and target
X = df.drop(columns=["Result"])
y = df["Result"]
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
# STEP 6: Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
# STEP 7: Create and train the model
model = ExtraTreesClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)
```

```python
model.fit(X_train, y_train)
```

```python
print("\nModel training completed!")
```

```python
# STEP 8: Predict test results
y_pred = model.predict(X_test)
```

```python
# STEP 9: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
```

```python
print(f"\nTest Accuracy: {accuracy * 100:.2f}%")
```

```python
# STEP 10: Detailed evaluation
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    labels=[-1, 1],
    target_names=["Phishing", "Legitimate"],
    zero_division=0
))
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred, labels=[-1, 1]))
```

```python
# STEP 11: Cross-validation
cv_scores = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
```

```python
print("\n5-fold CV Accuracy:",
      f"{cv_scores.mean() * 100:.2f}%")
```

```python
# STEP 12: Save the trained model
joblib.dump(model, "phishing__model.pkl")
```
