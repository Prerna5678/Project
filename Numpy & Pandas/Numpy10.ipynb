```python
'''
Broadcasting in NumPy

Broadcasting means NumPy automatically makes arrays compatible for mathematical operations 
even if their shapes are different.

It helps perform operations without writing loops.
'''
```

```python
import numpy as np
```

```python
# Broadcasting with Scalar
'''
Explanation

5 is added to every element automatically.
'''

arr = np.array([10, 20, 30, 40])

result = arr + 5

print(result)
```

```python
# Broadcasting with Two Arrays
# Element-wise addition happens.
arr1 = np.array([1, 2, 3])
arr2 = np.array([10, 20, 30])

result = arr1 + arr2

print(result)
```

```python
# Broadcasting Different Shapes

arr1 = np.array([[1], [2], [3]])

arr2 = np.array([10, 20, 30])

result = arr1 + arr2

print(result)
```

```python
'''
How It Works

Shape of arrays:

arr1 shape = (3,1)
arr2 shape = (3,)


NumPy expands arrays automatically:

[[1],      [10 20 30]
 [2],  +   
 [3]]


Becomes:

[[11 21 31]
 [12 22 32]
 [13 23 33]]
'''
```

```python
'''
5. Broadcasting Rules

Two arrays are compatible when:

Dimensions are equal
OR
One dimension is 1
'''
```

```python
# Example with 15–20 Data
# Total elements = 5 × 4 = 20 values
arr1 = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

arr2 = np.array([
    10,20,30,40
])

result = arr1 + arr2

print(result)
```

```python
# 7. Broadcasting Error Example

arr1 = np.array([1,2,3])

arr2 = np.array([10,20])

print(arr1 + arr2)
```

```python
# 9. Check Shape

print(arr1.shape)
print(arr2.shape)
```

```python
'''
Summary

| Concept             | Meaning                           |
| ------------------- | --------------------------------- |
| Broadcasting        | Automatic shape matching          |
| Scalar Broadcasting | One value applied to all elements |
| Compatible Shapes   | Same size or one dimension = 1    |
| Benefit             | Faster and less code              |

'''
```
