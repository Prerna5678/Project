```python
'''
Pivot Table in Pandas

A Pivot Table is used to summarize and analyze data.

It helps to:

Group data
Calculate sum, mean, count, etc.
Create reports easily
'''
```

```python
import pandas as pd
```

```python
data = {
    'Name': [
        'Prerna','Rahul','Amit','Sneha','Karan',
        'Neha','Arjun','Pooja','Vikas','Riya',
        'Ankit','Simran','Rohit','Meena','Yash',
        'Kriti','Dev','Tina','Aryan','Nisha'
    ],

    'Department': [
        'IT','HR','IT','Finance','HR',
        'IT','Finance','IT','HR','Finance',
        'IT','HR','Finance','IT','HR',
        'Finance','IT','HR','Finance','IT'
    ],

    'City': [
        'Ahmedabad','Delhi','Mumbai','Pune','Surat',
        'Jaipur','Chennai','Kolkata','Bhopal','Indore',
        'Lucknow','Patna','Nagpur','Rajkot','Vadodara',
        'Goa','Hyderabad','Noida','Kanpur','Agra'
    ],

    'Salary': [
        25000,30000,35000,28000,40000,
        32000,45000,38000,29000,41000,
        36000,27000,50000,33000,39000,
        42000,46000,31000,37000,34000
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
'''
3. Basic Pivot Table
Average Salary Department-wise
'''

pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc='mean'
)

print(pivot)
```

```python
# 4. Sum of Salary

pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc='sum'
)

print(pivot)
```

```python
# 5. Count Employees

pivot = pd.pivot_table(
    df,
    values='Name',
    index='Department',
    aggfunc='count'
)

print(pivot)
```

```python
# 6. Multiple Aggregations

pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc=['sum', 'mean', 'max', 'min']
)

print(pivot)
```

```python
'''
7. Using Columns Parameter
Department vs City
'''

pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    columns='City',
    aggfunc='sum'
)

print(pivot)
```

```python
# 8. Fill Missing Values

pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    columns='City',
    aggfunc='sum',
    fill_value=0
)

print(pivot)
```

```python
# 9. Add Margins (Total)

pivot = pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc='sum',
    margins=True
)

print(pivot)
```

```python
'''

Syntax:-

pd.pivot_table(
    dataframe,
    values='column_name',
    index='group_column',
    columns='optional_column',
    aggfunc='mean'
)


Summary:-

| Parameter  | Meaning                    |
| ---------- | -------------------------- |
| values     | Column to calculate        |
| index      | Group rows                 |
| columns    | Group columns              |
| aggfunc    | sum, mean, count, max, min |
| fill_value | Replace NaN                |
| margins    | Add total row              |

'''
```
