```python
''' 
Arithmetic operation NumPy array
1. Addition (+)
2. Subtraction (-)
3. Multiplication (*)
4. Matrix Multiplication (@)
5. Division (/)
6. Floor Division (//)
7. Exponentiation (**)
8. Module (%)
9. Transpose
'''
```

```python
import numpy as np
```

```python
# 2d array
# 1. Addition (+) 

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = x + y
print(z)
```

```python
# 2. Subtraction (-)

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = y - x
print(z)
```

```python
# 3. Multiplication (*) 

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = x * y
print(z)
```

```python
# 4. Matrix Multiplication (@)

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = x @ y
print(z)
```

```python
# 5. Division (/)
# it will return float value 

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = y / x
print(z)
```

```python
# 6. Floor Division (//)

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = y // x
print(z)
```

```python
# 7. Exponentiation (**)
# Exponentiation means power (eg: 12**2=144)

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = y ** x
print(z)
```

```python
# 8. Module (%)
# reminder

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

z = y % x
print(z)
```

```python
# 9. Transpose
# row will convert column
# column will convert row

x = np.array([[1 , 2],
              [3 , 4],])

y = np.array([[11 , 12],
              [13 , 14]])

print(x.transpose())
```
