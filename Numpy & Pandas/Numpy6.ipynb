```python
'''
Sorting 1d array
1. np.sort() --> it will return a sorted copy of an array
2. np.argsort() --> it will return indices that would sort an array
3. ndarray.sort() --> use array name and sort it in-place
'''
```

```python
import numpy as np
```

```python
# ascending order of rows default
x = np.array([7, 2, 3, 9, 6])
y = np.sort(x)
print(y)
```

```python
# use silicing for reverse / descending order
x = np.array([7, 2, 3, 9, 6])
y = np.sort(x)[::-1]
print(y)
```

```python
x = np.array([7, 2, 3, 9, 6])
y = np.argsort(x)
print(y)
```

```python
x = np.array([7, 2, 3, 9, 6])
x.sort()
print(x)
```
