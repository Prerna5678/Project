```python
'''
Unique and Count Values in NumPy

NumPy provides functions to find:

Unique values
Count of unique values
Frequency of each value
'''
```

```python
import numpy as np
```

```python
# 1. np.unique()

a = np.array([1,2,2,3,4,4,5])

result = np.unique(a)

print(result)
```

```python
# 2. Unique Values with Counts

a = np.array([1,2,2,3,4,4,4,5])

values, counts = np.unique(a, return_counts=True)

print("Unique Values:", values)
print("Counts:", counts)
```

```python
# 3. Loop Through Values and Counts

a = np.array([10,20,20,30,30,30])

values, counts = np.unique(a, return_counts=True)

for v, c in zip(values, counts):
    print(v, "appears", c, "times")
```

```python
# 4. Count Specific Value

a = np.array([1,2,2,2,3,4])

count = np.sum(a == 2)

print(count)
```

```python
# 5. Count Non-Zero Values

a = np.array([0,1,2,0,3,0,4])

count = np.count_nonzero(a)

print(count)
```

```python
# 6. Unique Values in 2D Array

a = np.array([[1,2],
              [2,3]])

result = np.unique(a)

print(result)
```

```python
'''
Important Functions

| Function             | Meaning               |
| -------------------- | --------------------- |
| `np.unique()`        | Find unique values    |
| `return_counts=True` | Count frequency       |
| `np.sum(a==x)`       | Count specific value  |
| `np.count_nonzero()` | Count non-zero values |

'''
```

```python
import numpy as np

a = np.array([5,10,10,15,15,15,20])

values, counts = np.unique(a, return_counts=True)

print(values)
print(counts)
```
