```python
'''
Stacking in NumPy

Stacking means:

👉 Combining multiple arrays into one array.

NumPy provides:

hstack() → horizontal stack
vstack() → vertical stack
dstack() → depth stack
'''
```

```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
'''
Horizontal Stacking (hstack)
Meaning
Join arrays side by side.
'''

np.hstack((a, b))
```

```python
'''
Vertical Stacking (vstack)
Meaning
Join arrays row-wise.
'''

np.vstack((a, b))
```

```python
'''
Depth Stacking (dstack)
Used mostly in 3D arrays/images.
'''

np.dstack((a, b))
```

```python
# Using concatenate()

np.concatenate((a, b))
```

```python
'''
2. Splitting in NumPy
Splitting means:

👉 Breaking one array into multiple parts.

Functions:

split()
hsplit()
vsplit()
'''
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])
```

```python
# Meaning
# Split array into 3 equal parts.

np.split(arr, 3)
```

```python
# Horizontal Split

arr2 = np.array([[1, 2, 3, 4],
                 [5, 6, 7, 8]])

np.hsplit(arr2, 2)
```

```python
# Vertical Split

np.vsplit(arr2, 2)
```

```python
'''
| Function   | Work                |
| ---------- | ------------------- |
| `hstack()` | join horizontally   |
| `vstack()` | join vertically     |
| `split()`  | divide array        |
| `hsplit()` | horizontal division |
| `vsplit()` | vertical division   |

'''
```
