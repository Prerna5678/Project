```python
'''
The random module in NumPy is used to generate random numbers, 
random arrays, shuffle data, and perform random sampling.
'''
```

```python
import numpy as np
```

```python
'''
Common Functions in NumPy Random Module
1. rand()

Generates random numbers between 0 and 1.

Example: 1D Array
'''

a = np.random.rand(5)
print(a)
```

```python
# Example: 2D Array

a = np.random.rand(2,3)
print(a)
```

```python
# 2. randint()

# Syntax: -np.random.randint(start, end, size)

a = np.random.randint(1, 10, 5)
print(a)
```

```python
'''
3. randn()

Generates random numbers from normal distribution.
'''

a = np.random.randn(5)
print(a)
```

```python
# 4. choice()

a = np.random.choice([10,20,30,40], 3)
print(a)
```

```python
# 5. seed()

np.random.seed(1)

print(np.random.rand(3))
```

```python
'''
6. shuffle()

Shuffles original array.
'''

a = np.array([1,2,3,4,5])

np.random.shuffle(a)

print(a)
```

```python
'''
7. permutation()

Returns shuffled array without changing original array.
'''

a = np.array([1,2,3,4,5])

b = np.random.permutation(a)

print(b)
print(a)
```

```python
'''
Difference Between shuffle() and permutation()

| Function      | Original Array Changes? |
| ------------- | ----------------------- |
| shuffle()     | Yes                     |
| permutation() | No                      |


Important Notes
rand() → decimal numbers
randint() → integer numbers
randn() → normal distribution values
choice() → random selection
shuffle() → changes original array
seed() → fixed random output
'''
```

```python
import numpy as np

# random integers
a = np.random.randint(1, 100, 10)

print(a)

# shuffle
np.random.shuffle(a)

print("After shuffle:", a)
```
