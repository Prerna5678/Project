```python
# Indexing of Numpy Array
```

```python
import numpy as np
```

```python
list1 = [10, 20, 30, 40, 50]
array1 = np.array(list1)
#print(array1)
print(array1[0])
```

```python
list1 = [10, 20, 30, 40, 50]
array1 = np.array(list1)
#print(array1)
print(array1[-1])
```

```python
list1 = [10, 20, 30, 40, 50]
array1 = np.array(list1)
#print(array1)
print(array1[2])
```

```python
list1 = [10, 20, 30, 40, 50]
array1 = np.array(list1)
#print(array1)
print(array1[-2])
```

```python
#          0    1    2 --> column
list1 = [[100, 200, 300] , # 0 rows
         [400, 500, 600],  # 1 rows
         [700, 800, 900]]  # 2 rows
array1 = np.array(list1)
#print(array1)
print(array1[0,1])
```

```python
#          0    1    2 --> column
list1 = [[100, 200, 300] , # 0 rows
         [400, 500, 600],  # 1 rows
         [700, 800, 900]]  # 2 rows
array1 = np.array(list1)
#print(array1)
print(array1[0,:])
```

```python
#          0    1    2 --> column
list1 = [[100, 200, 300] , # 0 rows
         [400, 500, 600],  # 1 rows
         [700, 800, 900]]  # 2 rows
array1 = np.array(list1)
#print(array1)
print(array1[:,1])
```
