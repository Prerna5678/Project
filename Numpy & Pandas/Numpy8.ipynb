```python
# Boolean indexing
```

```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr)
```

```python
arr > 25
```

```python
# Multiple Conditions
# AND condition

arr[(arr > 5) & (arr < 20)]
```

```python
# OR condition

arr[(arr < 10) | (arr > 20)]
```

```python
# Boolean Mask
# The condition array: is called a Boolean Mask.

arr > 10
```

```python
# 2D Array Example

arr = np.array([[10, 20],
                [30, 40]])

print(arr[arr > 20])
```
