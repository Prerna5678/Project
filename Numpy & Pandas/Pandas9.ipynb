```python
'''
String operations

Pandas provides .str functions to perform string operations on columns.

Used for:

Uppercase / lowercase
Replace text
Split strings
Find length
Remove spaces
Check words
etc.
'''
```

```python
import pandas as pd 
```

```python
data = {
    'Name': [
        'Prerna Baid',
        'Rahul Sharma',
        'Amit Patel',
        'Sneha Shah',
        'Karan Mehta',
        'Neha Joshi',
        'Arjun Singh',
        'Pooja Verma',
        'Vikas Yadav',
        'Riya Kapoor',
        'Ankit Jain',
        'Simran Kaur',
        'Rohit Gupta',
        'Meena Das',
        'Yash Modi',
        'Kriti Nair',
        'Dev Kumar',
        'Tina Roy',
        'Aryan Khan',
        'Nisha Patel'
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 3. Convert to Uppercase
df['Upper_Name'] = df['Name'].str.upper()

print(df)
```

```python
# 4. Convert to Lowercase

df['Lower_Name'] = df['Name'].str.lower()

print(df)
```

```python
# 5. Find Length of String

df['Length'] = df['Name'].str.len()

print(df)
```

```python
# 6. Replace Text

df['New_Name'] = df['Name'].str.replace('Patel', 'Patidar')

print(df)
```

```python
# 7. Remove Spaces

df['Clean_Name'] = df['Name'].str.strip()

print(df)
```

```python
# 8. Check Starting Letter

df['Starts_With_P'] = df['Name'].str.startswith('P')

print(df)
```

```python
# 9. Check Ending Word

df['Ends_With_Shah'] = df['Name'].str.endswith('Shah')

print(df)
```

```python
# 10. Split String

df[['First_Name', 'Last_Name']] = df['Name'].str.split(' ', expand=True)

print(df)
```

```python
# 11. Contains Word

df['Contains_Patel'] = df['Name'].str.contains('Patel')

print(df)
```

```python
# 12. Capitalize First Letter

df['Capitalize'] = df['Name'].str.capitalize()

print(df)
```

```python
# 13. Title Case

df['Title'] = df['Name'].str.title()

print(df)
```

```python
# 14. Count Characters

df['Count_a'] = df['Name'].str.count('a')

print(df)
```

```python
# 15. Extract Specific Text

df['Last_Name'] = df['Name'].str.extract(r'(\w+)$')

print(df)
```

```python
'''
17. Important .str Functions

| Function            | Work          |
| ------------------- | ------------- |
| `.str.upper()`      | Uppercase     |
| `.str.lower()`      | Lowercase     |
| `.str.len()`        | Length        |
| `.str.replace()`    | Replace text  |
| `.str.strip()`      | Remove spaces |
| `.str.split()`      | Split string  |
| `.str.contains()`   | Find word     |
| `.str.startswith()` | Check start   |
| `.str.endswith()`   | Check end     |

'''
```

```python
# 18. Complete Example

df = pd.DataFrame({
    'Name': [' Prerna Baid ', 'Rahul Sharma', 'Amit Patel']
})

df['Upper'] = df['Name'].str.upper()
df['Lower'] = df['Name'].str.lower()
df['Length'] = df['Name'].str.len()
df['Clean'] = df['Name'].str.strip()

print(df)
```

```python
'''
.str.strip()
means:

df['Name'] → selects the Name column
.str → allows string operations on column values
.strip() → removes extra spaces from beginning and end
df['Clean'] → stores cleaned result in new column called Clean

What strip() Removes

| Function    | Work                           |
| ----------- | ------------------------------ |
| `.strip()`  | Removes spaces from both sides |
| `.lstrip()` | Removes left-side spaces       |
| `.rstrip()` | Removes right-side spaces      |

'''
```

```python
text = "   hello   "

print(text.strip())   # both sides
print(text.lstrip())  # left side
print(text.rstrip())  # right side
```
