```python
'''
Correlation Matrix in Pandas

A correlation matrix shows the relationship between numerical columns.

It tells:

How strongly columns are related
Positive relationship
Negative relationship
No relationship

Values range from:

Value	Meaning
1	Perfect positive correlation
-1	Perfect negative correlation
0	No correlation
'''
```

```python
import pandas as pd
```

```python
data = {
    'Maths': [
        90,85,78,92,88,
        76,95,89,84,91,
        73,87,80,94,82,
        79,96,81,86,93
    ],

    'Science': [
        88,80,75,90,86,
        74,96,85,82,89,
        70,84,79,93,81,
        77,97,78,85,92
    ],

    'English': [
        85,78,72,88,84,
        70,92,80,76,87,
        68,82,75,90,79,
        73,94,74,81,89
    ],

    'Computer': [
        95,90,85,97,92,
        80,99,94,88,96,
        78,91,84,98,86,
        83,100,87,93,97
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 3. Find Correlation Matrix

corr_matrix = df.corr()

print(corr_matrix)
```

```python
'''
4. Meaning of Results
+ Positive Correlation
0.99

Means:

If Maths increases
Computer marks also increase

Strong positive relationship.

+ Negative Correlation
-0.85

Means:

One column increases
Other decreases

+ Zero Correlation
0

Means no relationship.

'''
```

```python
# 5. Select Specific Columns

df[['Maths', 'Science']].corr()
```

```python
# 6. Correlation with One Column

df.corr()['Maths']
```

```python
'''
8. Important Note

corr() works only on numerical columns.

If text columns exist:
'''

df.corr(numeric_only=True)
```

```python
# 9. Heatmap Visualization (Optional)

import matplotlib.pyplot as plt

corr = df.corr()

plt.imshow(corr)

plt.colorbar()

plt.show()
```

```python
# 10. Complete Example

data = {
    'Maths': [90, 80, 70, 60],
    'Science': [88, 78, 68, 58],
    'English': [85, 75, 65, 55]
}

df = pd.DataFrame(data)

print(df.corr())
```
