```python
# Sorting 2d array
```

```python
import numpy as np
```

```python
# default row wise sorting order follow

x = np.array([[12, 11, 15],
                [21, 25, 20],
                [18, 27, 16]])
y = np.sort(x)
#print(x)
print(y)
```

```python
# coloumn wise sorting order follow
# column axis=0 and row axis=1

x = np.array([[12, 11, 15],
                [21, 25, 20],
                [18, 27, 16]])
y = np.sort(x, axis=0)
#print(x)
print(y)
```

```python
# argument concept it will return indexing value only
# rows wise sort 
#        0    1   2
# eg:-  [12, 11, 15] ans:[1, 0, 2]
#              0    1   2
x = np.array([[12, 11, 15],
                [21, 25, 20],
                [18, 27, 16]])
y = np.argsort(x, axis=1)
#print(x)
print(y)
```

```python
# column wise sort argument

x = np.array([[12, 11, 15],
                [21, 25, 20],
                [18, 27, 16]])
y = np.argsort(x, axis=0)
#print(x)
print(y)
```

```python
# in-place

x = np.array([[12, 11, 15],
                [21, 25, 20],
                [18, 27, 16]])
x.sort()
print(x)
```
