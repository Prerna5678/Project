```python
''' 
Apply function 
apply() Function in Pandas

apply() is used to:

👉 Apply a function on rows or columns of a DataFrame.

Very useful for:

calculations
transformations
custom operations
'''
```

```python
import pandas as pd
```

```python
df = pd.read_csv('Housing.csv')
```

```python
df
```

```python
# Example 1: Increase Price by 10%
'''
Meaning

For every value in price column:

take value (x)
multiply by 1.10
'''

df['price'] = df['price'].apply(lambda x: x * 1.10)

print(df['price'])
```

```python
# Example 2: Convert Text to Uppercase

df['mainroad'] = df['mainroad'].apply(str.upper)

print(df['mainroad'])
```

```python
# Example 3: Create Custom Function

def square(x):
    return x * x

df['area'] = df['area'].apply(square)

print(df['area'])
```

```python
# Example 4: Categorize Prices

def category(x):
    if x > 5000000:
        return "High"
    else:
        return "Low"

df['price_category'] = df['price'].apply(category)

print(df[['price', 'price_category']])
```

```python
# Most Common Real Example

import pandas as pd

df = pd.read_csv("housing.csv")

# increase price
df['price'] = df['price'].apply(lambda x: x + 1000)

print(df.head())
```
